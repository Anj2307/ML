```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Titanic-Dataset.csv")
```

```python
#box plot
sns.barplot(x=df['Pclass'],y=df['Fare'],hue=df['Sex'])
```

```python
sns.boxplot(x=df['Age'],y=df['Sex'],hue=df['Survived'])
```

Sistplot(Numerical - catagorical)

```python

import matplotlib.pyplot as plt

# Histogram + KDE for both Survived categories
sns.displot(
    data=df,
    x="Age",
    hue="Survived",   # Color split by Survived (0/1)
    kind="hist",      # Histogram
    kde=True,         # Add KDE curve
    multiple="layer", # Overlap histograms
    height=5,
    aspect=1.5
)

plt.show()
```
