Initial placeholder in numpy arrays

```python
# create a numpy array of zeros
x=np.zeros((4,5))
print(x)
```

```python
#create a numpy array of ones
y=np.ones((3,3))
print(y)
```

```python
# array of a particular value
z=np.full((5,4),5)
print(z)
```

```python
#create a identity matrix
a=np.eye(4) #no. of rows and columns are equal so only one argument
print(a)
```

```python
# create a numpy array with random values
b=np.random.random((3,4))
print(b)
```

```python
# Random integers  values array within a specific range
c=np.random.randint(1,100,(10,10))
print(c)
print(c[0][1])
```

```python
# array of evenly spaced value
d=np.linspace(10,30,5) #(range_start,range_end,no. of values)
print(d)
```

```python
# array of evenly spaced values--> specifying the step
e=np.arange(10,30,4) #steps method
print(e)
```

```python
# convert a list to a numpy array
list2=[10,20,20,30,50]

np_array=np.asarray(list2)
print(type(np_array))
print(np_array)
```

Array Manipulation

```python
array=np.random.randint(0,10,(2,3))
print(array)
print(array.shape)
```

```python
#transpose\
trans=np.transpose(array)
print(trans)
```

```python
#another way to find transpose
trans2=array.T
print(trans2)
```

```python
# reshaping a array
a=np.random.randint(0,10,(2,3))
```

```python
b=a.reshape(3,2)
print(b)
print(a)
```
